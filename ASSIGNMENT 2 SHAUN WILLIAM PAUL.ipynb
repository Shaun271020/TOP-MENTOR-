```python

library = []

def add_book():
    book_id = input("Enter Book ID: ")

    # Check duplicate ID
    for book in library:
        if book["Book ID"] == book_id:
            print("Book ID already exists!")
            return

    title = input("Enter Book Title: ")
    author = input("Enter Author Name: ")

    book = {
        "Book ID": book_id,
        "Title": title,
        "Author": author,
        "Status": "Available",
        "Student": ""
    }

    library.append(book)
    print("Book added successfully!")


def display_books():
    if not library:
        print("No books available.")
        return

    print("\n------ Library Books ------")
    for book in library:
        print(f"""
Book ID : {book['Book ID']}
Title   : {book['Title']}
Author  : {book['Author']}
Status  : {book['Status']}
Student : {book['Student']}
------------------------------""")

def search_book():
    print("\nSearch By")
    print("1. Book ID")
    print("2. Book Title")
    print("3. Author Name")

    choice = input("Enter choice: ")

    if choice == "1":
        value = input("Enter Book ID: ")
        key = "Book ID"

    elif choice == "2":
        value = input("Enter Book Title: ")
        key = "Title"

    elif choice == "3":
        value = input("Enter Author Name: ")
        key = "Author"

    else:
        print("Invalid Choice")
        return

    found = False

    for book in library:
        if book[key].lower() == value.lower():
            print(f"""
Book ID : {book['Book ID']}
Title   : {book['Title']}
Author  : {book['Author']}
Status  : {book['Status']}
Student : {book['Student']}
""")
            found = True

    if not found:
        print("Book not found.")

def update_book():
    book_id = input("Enter Book ID to update: ")

    for book in library:
        if book["Book ID"] == book_id:
            book["Title"] = input("Enter New Title: ")
            book["Author"] = input("Enter New Author: ")
            print("Book updated successfully.")
            return

    print("Book not found.")

def delete_book():
    book_id = input("Enter Book ID to delete: ")

    for book in library:
        if book["Book ID"] == book_id:
            library.remove(book)
            print("Book deleted successfully.")
            return

    print("Book not found.")


def issue_book():
    book_id = input("Enter Book ID: ")

    for book in library:
        if book["Book ID"] == book_id:

            if book["Status"] == "Issued":
                print("Book already issued.")
                return

            student = input("Enter Student Name: ")
            book["Status"] = "Issued"
            book["Student"] = student

            print("Book issued successfully.")
            return

    print("Book not found.")

def return_book():
    book_id = input("Enter Book ID: ")

    for book in library:
        if book["Book ID"] == book_id:

            if book["Status"] == "Available":
                print("Book is already available.")
                return

            book["Status"] = "Available"
            book["Student"] = ""

            print("Book returned successfully.")
            return

    print("Book not found.")

def available_books():
    print("\nAvailable Books")

    found = False

    for book in library:
        if book["Status"] == "Available":
            print(f"""
Book ID : {book['Book ID']}
Title   : {book['Title']}
Author  : {book['Author']}
""")
            found = True

    if not found:
        print("No available books.")

def issued_books():
    print("\nIssued Books")

    found = False

    for book in library:
        if book["Status"] == "Issued":
            print(f"""
Book ID : {book['Book ID']}
Title   : {book['Title']}
Author  : {book['Author']}
Issued To : {book['Student']}
""")
            found = True

    if not found:
        print("No books are currently issued.")

while True:

    print("\n========== Library Management System ==========")
    print("1. Add New Book")
    print("2. Display All Books")
    print("3. Search Book")
    print("4. Update Book")
    print("5. Delete Book")
    print("6. Issue Book")
    print("7. Return Book")
    print("8. Display Available Books")
    print("9. Display Issued Books")
    print("10. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_book()

    elif choice == "2":
        display_books()

    elif choice == "3":
        search_book()

    elif choice == "4":
        update_book()

    elif choice == "5":
        delete_book()

    elif choice == "6":
        issue_book()

    elif choice == "7":
        return_book()

    elif choice == "8":
        available_books()

    elif choice == "9":
        issued_books()

    elif choice == "10":
        print("Thank you for using Library Management System.")
        break

    else:
        print("Invalid choice! Please try again.")
```
